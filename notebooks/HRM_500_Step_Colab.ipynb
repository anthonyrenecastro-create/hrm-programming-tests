# HRM 500-Step Reference Run
This notebook demonstrates the sandbox HRM simulation workflow.

```python
from hrm.engine import HRMEngine, SimulationConfig
from hrm.mesh import SimplicialMesh
from hrm.visualization import plot_history, plot_mesh
```

```python
mesh = SimplicialMesh.from_regular_grid(width=8, height=8, seed=7)
engine = HRMEngine(mesh, SimulationConfig(steps=500, base_dt=0.002))
engine.run()
```

```python
plot_history(engine, 'results')
plot_mesh(mesh, 'results/final_manifold_state.png', title='Final State')
```
